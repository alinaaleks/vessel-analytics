### Создание искусственных ФИО

```python
import random

surname_roots = [
    "Лан", "Вел", "Ром", "Тар", "Зор", "Мир", "Кал", "Нев", "Ор", "Сар",
    "Дал", "Пен", "Кир", "Вар", "Гор", "Лес", "Бор", "Зал", "Рин", "Том",
    "Эр", "Юл", "Акс", "Вен", "Лир", "Сет", "Нор", "Кас", "Мор", "Зен"
]

surname_suffixes = [
    "ев", "ин", "ов", "ар", "ель", "ан", "ор", "ис", "ун",
    "ак", "ир", "ас", "ел", "он", "ур", "ян", "ий", "ес"
]

male_endings = ["ов", "ин", "ский", "ев"]
female_endings = ["ова", "ина", "ская", "ева"]

def generate_surname(gender="m"):
    root = random.choice(surname_roots)
    suffix = random.choice(surname_suffixes)

    if gender == "m":
        return root + suffix + random.choice(male_endings)
    else:
        return root + suffix + random.choice(female_endings)
```

```python
male_parts_1 = [
    "Але", "Дми", "Сер", "Ан", "Мак", "Ро", "Ил", "Вла",
    "Ни", "Па", "Ге", "Ол", "Кон", "Ти", "Фе", "Ви",
    "Ар", "Ев", "За", "Юр"
]

male_parts_2 = [
    "дар", "мит", "рис", "тон", "вар", "сей", "ким", "дор",
    "лан", "мир", "ней", "рил", "сен", "тор", "вин", "дор",
    "тас", "рон", "вер", "гей"
]

def generate_male_name():
    return random.choice(male_parts_1) + random.choice(male_parts_2)
```

```python
female_parts_1 = [
    "Ана", "Еле", "Мар", "Оли", "Нат", "Ири", "Све", "Вик",
    "Дар", "Пол", "Ксе", "Люд", "Соф", "Али", "Жан", "Вер",
    "Лео", "Ми", "Те", "Юли"
]

female_parts_2 = [
    "лина", "рия", "нета", "сия", "дина", "сана", "лена",
    "мира", "нара", "тана", "вина", "ета", "ра", "сия", "яна"
]

def generate_female_name():
    return random.choice(female_parts_1) + random.choice(female_parts_2)
```

```python
patronymic_roots = [
    "Алек", "Ива", "Дми", "Сер", "Ан", "Ник", "Пав", "Мак",
    "Его", "Ром", "Вла", "Кон", "Мих", "Арт", "Вик", "Оле"
]

suffix_m = ["евич", "ович", "ич"]
suffix_f = ["евна", "овна", "ична"]

def generate_patronymic(gender="m"):
    root = random.choice(patronymic_roots)
    return root + random.choice(suffix_m if gender == "m" else suffix_f)
```

### Генерация

```python
# уникальность
used_names = set()

def ensure_unique(name):
    if name not in used_names:
        used_names.add(name)
        return name

    i = 2
    new_name = f"{name}_{i}"
    while new_name in used_names:
        i += 1
        new_name = f"{name}_{i}"

    used_names.add(new_name)
    return new_name
```

```python
# случайный шум
import random
import string

def noise():
    return f"{random.choice(string.ascii_uppercase)}{random.randint(10,99)}"
```

```python
# форматирование - компания
def format_company(name):
    return name.upper()
```

```python
def format_fio(name, org_type):
    return f"{name} {org_type.upper()}"
```

```python
def generate_fio_m():
    fio = f"{generate_surname('m')} {generate_male_name()} {generate_patronymic('m')}"
    return fio
```

```python
def generate_fio_f():
    fio = f"{generate_surname('f')} {generate_female_name()} {generate_patronymic('f')}"
    return fio
```

```python
def generate_company(prefix_list, suffix_list, org_type):
    name = f"{random.choice(prefix_list)} {random.choice(suffix_list)} {noise()} {org_type}"
    return format_company(name)
```

```python
def generate_name(row):

    t = row["type"]
    org = row["org_type"]

    # 👤 male fio
    if t == "fio_m":
        base = generate_fio_m()
        return ensure_unique(format_fio(base, org))

    # 👩 female fio
    if t == "fio_f":
        base = generate_fio_f()
        return ensure_unique(format_fio(base, org))

    # 🚛 freight / auto / agro / etc
    if t == "freight":
        return ensure_unique(generate_company(FREIGHT_PREFIX, FREIGHT_SUFFIX, org))

    if t == "auto":
        return ensure_unique(generate_company(AUTO_PREFIX, AUTO_SUFFIX, org))

    if t == "grain":
        return ensure_unique(generate_company(GRAIN_PREFIX, GRAIN_SUFFIX, org))

    if t == "terminal":
        return ensure_unique(generate_company(TERMINAL_PREFIX, TERMINAL_SUFFIX, org))

    if t == "facilities":
        return ensure_unique(generate_company(FACILITY_PREFIX, FACILITY_SUFFIX, org))

    if t == "bank":
        return ensure_unique(generate_company(BANK_PREFIX, BANK_SUFFIX, org))

    if t == "port_service":
        return ensure_unique(generate_company(PORT_SERVICE_PREFIX, PORT_SERVICE_SUFFIX, org))

    if t == "consult":
        return ensure_unique(generate_company(CONSULT_PREFIX, CONSULT_SUFFIX, org))

    if t == "survey":
        return ensure_unique(generate_company(SURVEY_PREFIX, SURVEY_SUFFIX, org))

    if t == "it":
        return ensure_unique(generate_company(IT_PREFIX, IT_SUFFIX, org))

    if t == "ins":
        return ensure_unique(generate_company(INS_PREFIX, INS_SUFFIX, org))

    if t == "gov":
        return ensure_unique(generate_company(GOV_PREFIX, GOV_SUFFIX, org))

    if t == "hotel":
        return ensure_unique(generate_company(HOTEL_PREFIX, HOTEL_SUFFIX, org))

    if t == "forward":
        return ensure_unique(generate_company(FORWARD_PREFIX, FORWARD_SUFFIX, org))

    if t == "storage":
        return ensure_unique(generate_company(STORAGE_PREFIX, STORAGE_SUFFIX, org))

    if t == "media":
        return ensure_unique(generate_company(MEDIA_PREFIX, MEDIA_SUFFIX, org))

    if t == "station":
        return ensure_unique(generate_company(STATION_PREFIX, STATION_SUFFIX, org))

    if t == "clean":
        return ensure_unique(generate_company(CLEAN_PREFIX, CLEAN_SUFFIX, org))

    if t == "customs":
        return ensure_unique(generate_company(CUSTOMS_PREFIX, CUSTOMS_SUFFIX, org))

    if t == "delivery":
        return ensure_unique(generate_company(DELIVERY_PREFIX, DELIVERY_SUFFIX, org))

    if t == "foreign":
        return ensure_unique(generate_company(FOREIGN_PREFIX, FOREIGN_SUFFIX, org))

    return ensure_unique(generate_company(FACILITY_PREFIX, FACILITY_SUFFIX, org))
```

```python
df["name_new"] = df.apply(generate_name, axis=1)
```

### Новые ИНН

```python
used_inns = set()

def generate_unique_inn(n_digits):
    while True:
        first = str(random.randint(1, 9))
        rest = "".join(str(random.randint(0, 9)) for _ in range(n_digits - 1))
        inn = first + rest

        if inn not in used_inns:
            used_inns.add(inn)
            return inn
```

```python
df["inn_new"] = df.apply(
    lambda row: generate_unique_inn(int(row["count_inn"])),
    axis=1
)
```
